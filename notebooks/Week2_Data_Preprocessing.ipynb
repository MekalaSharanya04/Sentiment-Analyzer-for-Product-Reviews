```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\Madhu\dataset\archive\amazon_reviews.csv")
```

```python
def get_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"

df["Sentiment"] = df["reviews.rating"].apply(get_sentiment)
```

```python
df[["reviews.text", "reviews.rating", "Sentiment"]].head()
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df[["reviews.text", "reviews.rating", "Sentiment"]].isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nSentiment Counts:")
print(df["Sentiment"].value_counts())
```

```python
df.dropna(subset=["reviews.text", "reviews.rating"], inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df.shape)
```

```python
df["Review"] = df["reviews.text"].astype(str).str.lower()
```

```python
df["Review"].head()
```

```python
import re

df["Review"] = df["Review"].apply(
    lambda x: re.sub(r"[^\w\s]", "", x)
)
```

```python
df["Review"] = df["Review"].str.replace(
    r"\d+", "", regex=True
)
```

```python
df["Review"] = df["Review"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)
```

```python
df["Review"].head()
```

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

stop_words = set(ENGLISH_STOP_WORDS)

def preprocess_text(text):
    # Tokenization
    words = text.split()
    
    # Remove stop words
    words = [word for word in words if word not in stop_words]
    
    # Simple lemmatization
    lemmatized_words = []
    
    for word in words:
        if word.endswith("ies"):
            word = word[:-3] + "y"
        elif word.endswith("ing") and len(word) > 5:
            word = word[:-3]
        elif word.endswith("ed") and len(word) > 4:
            word = word[:-2]
        elif word.endswith("s") and not word.endswith("ss") and len(word) > 3:
            word = word[:-1]
        
        lemmatized_words.append(word)
    
    return " ".join(lemmatized_words)

df["Clean_Review"] = df["Review"].apply(preprocess_text)
```

```python
df[["Review", "Clean_Review"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000)

X = tfidf.fit_transform(df["Clean_Review"])

print("TF-IDF Matrix Shape:", X.shape)
```

```python
y = df["Sentiment"]

print("Labels prepared successfully!")
print(y.head())
```
